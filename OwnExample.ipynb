```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
data = {
    "Study_Hours": [2, 3, 4, 5, 6, 7, 1, 8, 3, 6],
    "Attendance": [50, 60, 70, 80, 90, 95, 40, 98, 55, 85],
    "Assignments": [1, 2, 3, 4, 5, 5, 1, 5, 2, 4],
    "Result": [
        "Fail", "Fail", "Pass", "Pass", "Pass",
        "Pass", "Fail", "Pass", "Fail", "Pass"
    ]
}

df = pd.DataFrame(data)
df
```

```python
X = df.drop("Result", axis=1)
y = df["Result"]

print("Input Features:", X.columns.tolist())
print("Target Variable: Result")
print("Total Records:", len(df))

print("\nClass Counts:")
print(y.value_counts())
```

Implementing CART

```python
model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)

model.fit(X, y)

print("CART Model Trained Successfully")
```

```python
y_pred = model.predict(X)

print("Actual Results:", y.tolist())
print("Predicted Results:", y_pred.tolist())

accuracy = accuracy_score(y, y_pred)

print("Accuracy:", accuracy * 100, "%")
```

```python
plt.figure(figsize=(14, 8))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=model.classes_,
    filled=True,
    rounded=True
)

plt.title("CART Decision Tree - Student Pass/Fail Prediction")
plt.show()
```

```python
new_student = pd.DataFrame({
    "Study_Hours": [5],
    "Attendance": [75],
    "Assignments": [4]
})

prediction = model.predict(new_student)

print("Predicted Result:", prediction[0])
```

Summary:

In this assignment, I implemented the CART algorithm using a student performance dataset. I used Gini impurity to classify students as Pass or Fail based on study hours, attendance, and assignments. I trained the decision tree, evaluated its accuracy, visualized the tree, and predicted the result for a new student.

Result: The CART algorithm successfully classified students based on their academic performance.
